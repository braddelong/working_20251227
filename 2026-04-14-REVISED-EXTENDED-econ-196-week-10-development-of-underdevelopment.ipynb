## Review

### The End of the Agrarian & the Coming of the Imperial-Commercial Age

We left the agrarian age in three great steps. 

The first came around 1600, with what I like to call the commercial‑imperial age—or, just as well, perhaps better, the imperial‑commercial age. Europeans perfected ocean‑going ships that could reliably cross the Atlantic and Indian Oceans without sinking and could carry not just people, but bulk cargo.

the invention and diffusion of the ocean‑going caravel—the sturdy, maneuverable sailing ship of the Portuguese and Spanish empires—blew open the map. It made long‑distance Atlantic and Indian Ocean voyages routine rather than heroic one‑offs, and so underpinned the first great wave of global trade, colonization, and resource extraction. 

Those ships bore two very different kinds of freight. On the one hand, they carried the “useful goods that people wanted to buy”: spices, sugar, cotton textiles, coffee, tea, silver. On the other hand, they carried organized violence: cannon and soldiers. The combination of cheap long‑distance shipping and mobile firepower transformed what states could do. It opened up new opportunities for trade, for a finer division of labor across continents, and for what we would now call “biotechnology exchange”—the spread of crops, animals, and germs. It also opened up unprecedented opportunities for domination: sending your people somewhere far away, coercing or taxing the locals, and making them ship resources back to you.

In practice, durable, large‑scale, fiscally meaningful extraction at a distance turned out to be hard. If you look closely, there is really only one place in the early modern world where an imperial power consistently managed to tax a conquered population and send truly world‑historical quantities of resources back across the ocean: Potosí, in what is now southern Bolivia (at the time part of the Spanish Viceroyalty of Peru).

At Potosí, there was essentially a mountain of silver—the Cerro Rico, or “rich hill.” Spanish conquistadors and their descendants dismantled that mountain over the course of centuries, mining enormous quantities of silver ore. The silver was refined and shipped to Spain. The Spanish crown and its grandees then used this silver hoard to buy goods across Europe and Asia and, crucially, to fund armies and fleets in the wars of the Counter‑Reformation. The Spanish fiscal‑military state, and its attempts to “re‑Catholicize” or at least dominate large chunks of Europe, quite literally ran on Bolivian silver. That flow was world‑historically significant. (On Potosí and Spanish silver, see, e.g., Dennis Flynn and Arturo Giráldez, “Born with a ‘Silver Spoon’: The Origin of World Trade in 1571,” Journal of World History 6(2), 1995; or Kris Lane, Potosí: The Silver City That Changed the World, University of California Press, 2019.)

Outside that exceptional case, the biggest resource transfers associated with the commercial‑imperial age took two other forms.

First, there was settler colonialism. Europeans and their descendants moved to temperate lands—North America, the Southern Cone of South America, Australasia, parts of southern Africa—and established farms, then towns, then cities, and eventually manufacturing centers. These “neo‑Europes” transformed land that had previously been lightly farmed, pastoral, or used under very different social arrangements into an extension of Europe’s agricultural and industrial base. This is what historians like James Belich have in mind when they talk about the “settler revolution” and the creation of an “Anglo‑world” spanning multiple continents.

Second, and more sinister, the new ships made the Atlantic slave system vastly more profitable and scalable. Before about 1600, there were certainly slave trades around the world. People were kidnapped, sold, and worked to death in many places. What changes with the coming of robust, ocean‑crossing sailing ships is the sheer scale and distance over which slavery is organized.

Once you can move cargo cheaply across the Atlantic, it becomes worthwhile for Europeans to sail to the West African coast and trade guns and manufactured goods to local rulers in exchange for enslaved people. If you are an African king on that coast and you refuse this bargain, you face a terrible choice: your neighbors, who eagerly accept European guns, now have more firepower than you. Very soon you are not a king at all; you are an ex‑king, quite possibly enslaved yourself. Caught in this arms race, coastal elites raid the interior for captives, bring them to the coast, and sell them to European slavers.

European ships then carry these enslaved Africans across the Atlantic, primarily to the Caribbean and parts of Brazil. There they are forced to grow sugar and tobacco on plantations. Those commodities are shipped to Europe, where they feed a rapidly growing demand for cheap stimulants and luxuries. In return, guns, metalware, textiles, and other goods go back to the African coast. This is the classic “triangular trade”:

- Europe → Africa: guns and manufactures

- Africa → Americas: enslaved people

- Americas → Europe: sugar, tobacco, later coffee, cotton, and more

The result is an enormous amount of wealth generated in plantation zones and European ports; but it is wealth built on coerced labor, high mortality, and the militarization of African states. The commercial‑imperial age thus ushers in a “society of domination” on a new scale: empires and plantation complexes able to project violence and extract labor and resources across oceans. (For classic treatments, see Eric Williams, Capitalism and Slavery (1944), and more recent overviews like Sven Beckert, Empire of Cotton (Knopf, 2014).)

To be clear, none of this means that slavery or predation were new in 1600. What is new is the technology and organization that make these systems truly global. Better ships, better navigation, better military organization, and more sophisticated fiscal states allow Europeans to integrate far‑flung regions into what we would now recognize as a single world economy. That is why I label this first great step out of the agrarian age a watershed: it dramatically enlarges both the possibilities for mutually beneficial trade and the capacities for organized theft—and it does so on a genuinely planetary scale.

### The British Industrial Revolution Century

Another real discontinuity comes with 1775. For the first time, we begin to deploy serious non‑animal power sources—something other than the horse or the ox—at scale. That is the British Industrial Revolution. What changes is not that people discover coal; people have been digging up coal and burning it to keep warm for a very long time. What changes is that people finally figure out how to use coal to do work: to move things, to create pressure, to power machines.

Coal is, in an important sense, solar power. It is just solar power from a very long time ago. Some 300 to 350 million years back—“Carboniferous,” not “700 million,” if we are being geologically pedantic—the sun shone on the earth. Plants used that sunlight to grow. A lot of that plant matter then died, fell into swamps, and was buried under mud and rock. Over tens of millions of years, under heat and pressure, that buried organic matter was transformed from wood and leaves into coal. Later, we humans dug that concentrated fossil carbon up and burned it. In doing so, we tapped stored sunlight—solar energy sequestered in the deep past—to power the machines of the Industrial Revolution: textile machinery in the new factories, high‑pressure pumps and steam engines, iron smelters, and, soon enough, railroads.

The domestication of the horse and the ox had been a very big deal. It let humans deploy far more force than their own muscles could provide—to plow heavier soils, haul larger loads, move more goods over longer distances. But as important as draft animals were, the arrival of coal as a power source was a much bigger deal.

Coal broke the biological ceiling. Horses and oxen are ultimately limited by how much fodder you can grow with current sunlight and how much stress an animal’s body can take. Coal, by contrast, is stored sunlight from the deep past: plant matter that grew hundreds of millions of years ago, died, was buried, and under geological heat and pressure turned into concentrated carbon. Once people learned not just to burn coal for heat, but to harness it to do mechanical work, the constraint on power shifted from biology to geology and engineering.

This shift is what underlies the sharp upward jump in technology in the late eighteenth and early nineteenth centuries.

Around 1775, we get the clustered, mutually reinforcing innovations we conventionally bundle together as the “Industrial Revolution”: the steam engine, the spinning machines (from Hargreaves’s spinning jenny to Arkwright’s water frame to Crompton’s mule), the power loom or weaving machine, the new iron‑smelting technologies (notably coke‑fired blast furnaces), and, soon after, the railroad as an integrated system of steam locomotives, iron rails, and scheduled mass transport. That is not really a “quadruple” invention; it is at least five major, tightly linked technologies that reorganize energy use, manufacturing, and transportation all at once.

And then in 1875 we get something qualitatively different: not just another upward jump in technology, but a full‑scale phase change in how the economy works.

### The Post-1875 Modern Economic Growth Era

After 1875 the innovations no longer come as isolated clusters separated by a century or more. Instead, a self‑sustaining innovation system emerges, centered in what contemporaries called the “Second Industrial Revolution” or the “age of steel, electricity, and chemicals.” You get large, professional R&D labs; systematic application of science (especially physics and chemistry) to industry; patent systems and corporate forms that scale; and education systems that reliably produce engineers and technicians. From then on, technological progress is not a series of discrete jumps but a continuous avalanche. The structure of growth changes.

If the 1600 caravel revolution is about extending the geographic reach of trade, and the 1775 cluster is about mechanizing production and transportation with steam and iron, then the post‑1875 world is about building a permanent engine for invention itself. Electricity, the internal combustion engine, the telegraph and telephone, synthetic fertilizers and chemical dyes, modern steelmaking, and later on mass production and the assembly line: these are not just new gadgets, they are manifestations of a new regime in which innovation is institutionalized and expected. The global economy ceases slouching forward via occasional, localized breakthroughs, and starts racing ahead under the constant pressure of organized research, industrial competition, and state policy.

That is why 1875 belongs in a different mental category. 1600 and 1775 give us powerful new tools. After 1875, we start systematically reinventing the toolkit itself, decade after decade.

Technological progress as we know it really starts to look different around 1875. At that point, we don’t just get new gadgets; we get the invention of systematic innovation—a durable, institutional machinery for generating and applying new ideas over and over again.

What makes this possible is the coming-together of the "four-in-hand" institutions:

First, there is modern science as we would recognize it today. By the late nineteenth century, experimental science has reached critical mass. Instead of a scattering of lone geniuses, we have organized, professional communities: laboratories, journals, universities, and scientific societies producing reliable knowledge at scale. This is the world of Maxwell, Pasteur, and their peers, in which experiment, theory, and replication interact inside durable institutions to make discovery a regular process rather than an occasional miracle.

Second, the industrial research laboratory appears as a bridge from science to engineering. Firms stop relying only on independent tinkerers and instead build in-house labs whose job is to turn “pure” scientific understanding into usable devices, processes, and materials. Thomas Edison’s Menlo Park, and later Bell Labs and General Electric’s research labs, are archetypes here: they are factories for innovation, systematically searching for commercially valuable ways to harness scientific laws and principles. Jon Gertner’s The Idea Factory is an excellent narrative of how Bell Labs did this in the twentieth century, and Alfred Chandler’s work on the rise of industrial research inside large firms shows how deeply this logic penetrated modern business.

Third, you get the modern corporation, which is not just a big workshop but an organizational technology for scaling up production across space. A modern corporation can take the products of the engineers and deploy them not in one factory or one town but across a continent or around the globe. It can build many plants, all using the same knowledge, designs, and processes, and coordinate them via managerial hierarchies, accounting systems, and standardized procedures. Chandler’s The Visible Hand is the canonical account of how railroads, steel, chemicals, and later autos pioneered this form and why it mattered for growth.

Fourth, there is the globalized market economy, which makes it profitable to do all this at scale. Falling transport and communication costs, expanding international trade, and integrated capital markets create a world market for new products and processes. If a research lab and a corporation can come up with a better way to make steel, generate electricity, refine chemicals, or process food, there is a worldwide demand large enough to justify the fixed costs of research, plant construction, and organizational complexity. Kevin O’Rourke and Jeffrey Williamson’s Globalization and History tells the story of how, from the late nineteenth century on, world markets knit together into a single integrated system.

Put these four together—critical-mass science, the industrial research lab, the modern corporation, and the global market—and you get a self-reinforcing engine. Science produces knowledge; labs turn knowledge into usable technologies; corporations scale those technologies into standardized, mass production; global markets reward successful deployments with profits that can then be plowed back into the next round of research and expansion. The outcome is the world we inhabit today, in which underlying technological capacity grows at on the order of 2 percent per year, rather than 2 percent per half-century as in the pre-1870 world.

A 2 percent annual growth rate sounds modest in any given year, but compounded over decades it is the difference between a society that looks much like its grandparents’ and one in which the entire material and organizational landscape is transformed beyond recognition. That shift brings with it immense opportunities: to cure disease, extend lifespans, reduce drudgery, and generate a level of material abundance that would have struck our pre-industrial ancestors as literal fantasy. It also brings immense dangers. The same machinery that gives us antibiotics, microchips, and container ships also gives us high explosives, weapons of mass destruction, and surveillance-and-control technologies on a planetary scale.

The system of systematic innovation that crystallizes after 1875, in other words, is not a simple “march to utopia.” It is an accelerator—one that speeds up both our capacity to do good and our capacity to destroy ourselves and our world. Whether that 2 percent per year trend ends up being a blessing or a curse depends not just on the institutions of innovation, but on the institutions of governance and restraint that we manage to build around them.

For a fuller exploration of how this post-1870 regime of rapid growth emerged, what it did, and why it failed to deliver us to a genuine utopia, see my own Slouching Towards Utopia: An Economic History of the Twentieth Century (Basic Books edition (http://bit.ly/3pP3Krk)).

### The Agricultural Surplus Drives Everything

* The First World farm family producing 12,000 lbs has ~8,000–10,000 lbs above subsistence to sell, invest, or be taxed.
* That surplus funds: railways, schools, machinery, banks, urban infrastructure, and the domestic market that makes industrialization viable. 
* **Industrialization requires a home market: you learn to manufacture by selling in a protected home market first, and a home market requires farm families with purchasing power above subsistence.**
* The tropical countries were offered trade instead of industrialization—and took it, because trade was immediately profitable given their labor costs: exporting cotton, tea, rubber, and cocoa generated cash income above the subsistence baseline but locked them into the structural trade pattern.
* Demand for primary products relatively inelastic -> technological progress lowers prices, with extra surplus going to users-consumers.
* Demand for secondary (industrial) products relatively elastic -> technological progress increases quantity, with extra surplus going to sellers-producers.

### Migration Drove the Divide

* ~50 million Europeans migrated to temperate settlements (US, Canada, Australia, Argentina), bringing skills, institutions, and the expectation of First World wages.
* ~50 million Indians and Chinese migrated to tropical plantations as indentured labor, creating the low-wage equilibrium in tropical countries that were originally rich.
* The two migration streams went to different wage worlds and deepened them.
* Why couldn't Indian and Chinese workers move in large numbers to temperate economies? Racism.
* The trap is self-reinforcing:
    * Low farm productivity → low wages → cheap exports → no surplus → no home market → no industrialization → labor stays in low-productivity agriculture → even with technological progress in primary-product producing industries, surplus flowed to users-consumers → low wages persist.
    * The First World ran the same loop in reverse: high farm productivity → high wages → expensive exports → large surplus → big home market → industrialization → higher productivity → higher wages.

### Moving a Little Toward Generality

* $ \pi $ :: share of population in manufacturing
* $ s_m $ :: share of manufacturing in West
* $ s_n = (1-\pi)/2 + {\pi}s_m $ :: share of total population in West
* $ x $ :: sales of a typical manufacturing firm
* $ F $ :: fixed costs of a manufacturing plant
* $ t $ :: cost of shipping a unit of manufactures across regions

Then:

* cheaper to service West from East as long as: $ {s_n}xt < F $
* cheaper to service West from East as long as: $ (1-{s_n})xt < F $
* cheaper to have two plants otherwise

Algebra:

* $ s_m = 0 $ if $ s_n < \frac{F}{tx} $
* $ s_m = s_n $ if $ \frac{F}{tx} < s_n < 1 - \frac{F}{tx} $
* $ s_m = 1 $ if $ 1- \frac{F}{tx} < s_n $
