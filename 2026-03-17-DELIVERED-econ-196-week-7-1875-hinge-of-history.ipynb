## Modeling Rates of Technological Progress

It was, I think, in 1985 back when I was still a graduate student that I really became aware of the "two heads are better than one" argument as I watched Paul Romer present a piece of his dissertation—a piece that had already been accepted by and was about to be published in the _Journal of Political Economy_. 

* Romer, Paul M. 1986. “Increasing Returns & Long-Run Growth.” _Journal of Political Economy_ 94(5): 1002–37. <<https://paulromer.net/speeding-up-and-missed-opportunities-evidence/IncreasingReturns.pdf>>.
* Kremer, Michael. 1993. “Population Growth & Technological Change: One Million B.C. to 1990.” _Quarterly Journal of Economics_ 108 (3): 681–716. <<https://faculty.econ.ucdavis.edu/faculty/gclark/210a/readings/kremer1993.pdf>>>
* Diamond, Jared. 1997. _Guns, Germs, and Steel: The Fates of Human Societies_. New York: W. W. Norton. <<https://dn721503.ca.archive.org/0/items/fp_Jared_Diamond-Guns_Germs_and_Steel/Jared_Diamond-Guns_Germs_and_Steel.pdf>>.

Paul did something that, in retrospect, is stunningly simple: it took the old production function, pulled ideas out of the undifferentiated capital aggregate, and insisted that they were different—non-rival, accumulating, and capable of generating increasing returns at the social level even when private firms still faced diminishing returns.

Once you see that, you cannot unsee it. Non-rivalry means that if I have the blueprint for the Watt steam engine or for CRISPR, my using it does not prevent you from using it as well. We can both plug the same idea into our production lines, and the marginal cost of your adoption is, essentially, zero. A world made up only of rival goods—land, raw labor, iron ore—grinds against diminishing returns and the Malthusian devil. Add a stock of non-rival ideas that can be copied and recombined, and suddenly the growth rate of knowledge, ￼, can be written, in the simplest possible reduced form, as proportional to the number of brains available to think: ￼.

That is where Michael Kremer enters, walking onstage from the long run of one million B.C. to 1990. Kremer basically says: take Romer’s “two heads are better than one” literally and push it back nearly to Homo erectus with a sharp stick. If ideas are non-rival, more people mean more potential inventors. Combine that with a Malthusian world in which population is limited by technology, and you get the striking empirical regularity Kremer documents: for most of human history, the growth rate of population is roughly proportional to the level of population. More people → more ideas → higher carrying capacity → more people. A feedback loop, not a steady state.

Jared Diamond gives us the spatial counterpart to this temporal story. Geography, crops, domesticable animals, and continental axes determine where dense populations, cities, and written traditions can form. Those are precisely the institutions that let a society store, copy, and recombine non-rival ideas at scale. An Andean herder and a Fertile Crescent scribe are not different in cognitive capacity; what is different is how many other heads their ideas can bump up against, how many previous ideas they can draw on, and how cheaply those ideas can be transmitted.

Put Romer, Kremer, and Diamond together and you get a macrostructure of human history that looks very much like the hockey-stick plots in your notebook: for millennia, population, globa income, and a very crude tech index creep upward as the non-rival idea engine runs slowly in small, partially isolated populations; then the commercial-imperial and industrial eras abruptly knit those populations into a single “anthology intelligence” with vastly expanded means of R&D, and the growth rate of technology jumps by an order of magnitude or more—and keeps on growing.

The true scarce resource in the long run is not stuff; rather, it has been arrangements—non-rival patterns of information and organization that let us get more out of the same rival inputs. History’s hinge points are the moments when we change how many heads can share how many ideas how fast.

Now there is "stepping on toes": two heads are not twice as good as one

All that suggests a very simple model of technological change as a function of world population:

$$
\begin{equation}
\frac{\dot H_t}{H_t} =  \alpha P_t^\sigma
\end{equation}
$$

Where H is our index of technology—human ideas—and P our estimate of population, and alpha and sigma are parameters. 

Let's take sigma = 0.5.

Now you can immediately start complicating this "two heads are better than one" story further. But let's roll with this.

---
---

Start with one of the simplest possible fundamentalist Malthusian models of human history, with the equations of the model being:

* $ n = \beta \left( \frac{y - y_{sub}}{y_{sub}}  \right) $
* $ g = h - n/\gamma $
* $ h = \alpha P^{\sigma} $

And our definitions:

* $ \frac{1}{H} \frac{dH}{dt} = h $
* $ \frac{1}{y} \frac{dy}{dt} = g $
* $ \frac{1}{P} \frac{dP}{dt} = n $

That is:

* population growth *n* is an increasing function of productivity and standard-of-living *y* relative to subsistence $ y_{sub} $, with a parameter $ \beta $ controlling how fast population rises or falls as standard-of-living rises or falls above the subsistence level
* labor-productivity growth *g* is a positive function of technological progress *h* and a negative function of population growth *n*. Why the second? Because population growth puts more pressure on the fixed stock of natural resources. The relative salience of technology vis-à-vis resources is governed by a parameter $ \gamma $.
* technological progress *h* is an increasing function of human population *P*—because two heads are better than one—but a less than linear function—because of duplication of inventive effort, people stepping on one another's toes. The parameter $ \alpha $ governs how effective people are at pushing technology forward; the parameter $ \sigma $ captures the stepping-on-toes duplication of effort.
* *h* is the proportional growth rate of *H*, which is the value of the stock of "technology"—useful ideas about manipulating nature and coöperatively organizing humans.
* *g* is the proportional growth rate of *y*, which is the standard-of-living, income, productivity.
* *n* is the proportional growth rate of the human population *P*.

The model has five parameters: $ \alpha, \sigma, \gamma, \beta, y_{sub} $. They are the two-heads-are-better-than-one coefficient, the stepping-on-toes coefficient, the salience of ideas vis-à-vis resources in determining productivity, the annual response of population growth to relative living standards, and the Malthusian subsistence level of productivity. What values should we choose these parameters?

We set $ y_{sub} $ at 1100 2025-value dollars as annual average *per capita* income.

We set $ \beta  = 0.005 $: a doubling from subsistence of the standard-of-living induces a positive population growth rate of 0.5% per year.

We set $ \gamma = 2 $: ideas are twice as salient for prosperity as natural resources *per capita*.

We let $ \alpha $ and $ \sigma $ be whatever is needed to fit some section of the historical path of technological development.

And add in some random shocks for verisimilitude...

Let us then fit, or rather "fit"—it is not a statistical fit—this model; fit it over the period from the year -3000 development of writing, bronze, and the state to the year 200 classical-antiquity civilizational apogee; fit it to my finger-exercise database of guesstimates about the structure of human economic history in the very long run.

## The Malthusian Model

### Transforming Code into Functions

```python
# make a model as a function:

def simulate_history(
    human_history_df,
    alpha=0.000028,   # two-heads-are-better-than-one coefficient
    sigma=0.5,        # stepping-on-toes coefficient
    gamma=2,          # salience of ideas vis-à-vis resources in determining productivity
    beta=0.005,       # annual response of population growth to relative living standards
    y_sub=1100,       # subsistence level of productivity
    initial_year=-3000,
    final_year=2025
):
    """
    Run the Kremer/Romer-style simulation from initial_year to final_year.

    Returns:
        g, n, H, P, y, h, year
        each as a Python list, in chronological order from initial_year.
    """

    # initial conditions from the dataframe
    initial_H = human_history_df.loc[
        human_history_df["year"] == initial_year, "tech_index"
    ].iloc[0]
    initial_P = human_history_df.loc[
        human_history_df["year"] == initial_year, "population"
    ].iloc[0]
    initial_y = human_history_df.loc[
        human_history_df["year"] == initial_year, "income"
    ].iloc[0]

    # time-series containers
    g = []                       # productivity growth
    n = []                       # population growth
    epsilon = []                 # population shocks
    H = [initial_H]              # tech level
    P = [initial_P]              # population / labor force
    y = [initial_y]              # productivity / income
    h = [alpha * np.power(P[0], sigma)]  # tech growth rate
    year = [initial_year]        # calendar year

    # number of yearly steps
    T = final_year - initial_year

    for t in range(T):
        # draw this year's population shock
        epsilon_t = rnd.normalvariate(0, 0.01)
        epsilon.append(epsilon_t)

        # population growth: n_t
        n_t = beta * (y[t] - y_sub) / y_sub + epsilon_t
        n.append(n_t)

        # productivity growth: g_t
        g_t = h[t] - n_t / gamma + rnd.normalvariate(0, 0.0002)
        g.append(g_t)

        # next period technology growth rate h_{t+1}
        h_next = alpha * np.power(P[t], sigma)
        h.append(h_next)

        # update levels
        y_next = y[t] * np.exp(g_t)
        P_next = P[t] * np.exp(n_t)
        H_next = H[t] * np.exp(h[t])   # note: uses current h[t], as in your original
        y.append(y_next)
        P.append(P_next)
        H.append(H_next)

        # increment calendar year
        year.append(year[-1] + 1)

    return g, n, H, P, y, h, year
```

```python
# make the graphical presentation as a function

import numpy as np
import matplotlib.pyplot as plt

def safe_log(x):
    x = np.array(x, dtype=float)
    x[x <= 0] = np.nan
    return np.log(x)

def plot_sim_vs_history_multiples(human_history_df, year, P, y, H, h):
    """
    Plot 2×2 panels comparing simulation to historical data
    over three different time windows:

      1. -3000 to 200
      2. -3000 to 1775
      3.  200 to 2025
    """

    # Convert simulation lists to arrays
    year_arr = np.array(year)
    P_arr    = np.array(P)
    y_arr    = np.array(y)
    H_arr    = np.array(H)
    h_arr    = np.array(h)

    # Prepare historical dataframe (once)
    df = human_history_df.copy()
    df["year"] = df["year"].astype(int)

    # Convenience function to build one 2×2 figure for a given window
    def _make_figure(year_min, year_max,
                     ytick_income_levels,
                     title_suffix):
        mask   = (df["year"] >= year_min) & (df["year"] <= year_max)
        df_sub = df.loc[mask].copy()

        years_df = df_sub["year"].values
        pop_df   = df_sub["population"].values
        inc_df   = df_sub["income"].values
        tech_df  = df_sub["tech_index"].values
        gtech_df = df_sub["tech_growth"].values

        # restrict simulation to same window
        sim_mask = (year_arr >= year_min) & (year_arr <= year_max)
        year_sim = year_arr[sim_mask]
        P_sim    = P_arr[sim_mask]
        y_sim    = y_arr[sim_mask]
        H_sim    = H_arr[sim_mask]
        h_sim    = h_arr[sim_mask]

        lnP_sim = safe_log(P_sim)
        lny_sim = safe_log(y_sim)
        lnH_sim = safe_log(H_sim)

        lnP_data = safe_log(pop_df)
        lny_data = safe_log(inc_df)
        lnH_data = safe_log(tech_df)

        fig, axes = plt.subplots(2, 2, figsize=(14, 8))

        # (1,1) log P and log population vs year
        ax = axes[0, 0]
        ax.plot(year_sim, lnP_sim, color="tab:blue", label="ln P (simulated)")
        ax.plot(years_df, lnP_data, color="tab:orange", marker="o",
                linestyle="-", label="ln population (historical)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Population (millions)")
        yticks_level = np.array([1e1, 1e2, 1e3, 1e4])
        yticks_log   = np.log(yticks_level)
        ax.set_yticks(yticks_log)
        ax.set_yticklabels([f"{val:,.0f}" for val in yticks_level])
        ax.set_title(f"Log P vs log population ({year_min} to {year_max})")
        ax.grid(True, alpha=0.3)
        ax.legend()

        # (1,2) log y and log income vs year
        ax = axes[0, 1]
        ax.plot(year_sim, lny_sim, color="tab:blue", label="ln y (simulated)")
        ax.plot(years_df, lny_data, color="tab:orange", marker="o",
                linestyle="-", label="ln income (historical)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Income / productivity")
        yticks_log = np.log(ytick_income_levels)
        ax.set_yticks(yticks_log)
        ax.set_yticklabels([f"{val:,.0f}" for val in ytick_income_levels])
        ax.set_title(f"Log y vs log income ({year_min} to {year_max})")
        ax.grid(True, alpha=0.3)
        ax.legend()

        # (2,1) log H and log tech_level vs year
        ax = axes[1, 0]
        ax.plot(year_sim, lnH_sim, color="tab:blue", label="ln H (simulated)")
        ax.plot(years_df, lnH_data, color="tab:orange", marker="o",
                linestyle="-", label="ln tech_level (historical)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Log technology level")
        yticks_level = np.array([0.05, 0.1, 0.3, 1.0], dtype=float)
        yticks_log   = np.log(yticks_level)
        ax.set_yticks(yticks_log)
        ax.set_yticklabels([f"{val:,.2f}" for val in yticks_level])
        ax.set_title(f"Log H vs log tech_level ({year_min} to {year_max})")
        ax.grid(True, alpha=0.3)
        ax.legend()

        # (2,2) h and tech_growth vs year (levels)
        ax = axes[1, 1]
        ax.plot(year_sim, h_sim, color="tab:blue", label="h (simulated)")
        ax.plot(years_df, gtech_df, color="tab:orange", marker="o",
                linestyle="-", label="tech_growth (historical)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Technology growth")
        ax.set_title(f"h vs tech_growth ({year_min} to {year_max})")
        ax.grid(True, alpha=0.3)
        ax.legend()

        fig.suptitle(title_suffix, y=1.02, fontsize=16)
        plt.tight_layout()
        plt.show()

    # 1) -3000 to 200: early long-run
    _make_figure(
        year_min=-3000,
        year_max=200,
        ytick_income_levels=np.array([1000, 1250, 1500]),
        title_suffix="Simulation vs History, -3000 to 200"
    )

    # 2) -3000 to 1775: pre-Industrial takeoff
    _make_figure(
        year_min=-3000,
        year_max=1775,
        ytick_income_levels=np.array([1000, 1500, 2000, 2500, 3000]),
        title_suffix="Simulation vs History, -3000 to 1775"
    )

    # 3) 200 to 2025: “hockey-stick” era
    _make_figure(
        year_min=200,
        year_max=2025,
        ytick_income_levels=np.array([1000, 3000, 10000, 30000]),
        title_suffix="Simulation vs History, 200 to 2025"
    )
```

Here is a simulation run, where we let the Malthusian model run past the year 200 all the way up to the year 2025. The blue lines are our model simulation. The orange dots connected by straight lines are my guesstimate historical benchmarks, up to year 200:

```python
# run our functions...

# NEW run:
g, n, H, P, y, h, year = simulate_history(human_history_df)

# Plot that run
plot_sim_vs_history_multiples(human_history_df, year, P, y, H, h)
```

## Class Lecture Notes

March 17, 2026

Back in the mid‑1980s, people like Paul Romer, Michael Kremer, and others crystallized a very attractive view of long‑run growth. Ideas are non‑rival: once one brain figures something out, other brains can copy it at essentially zero marginal cost. Humans are hyper‑communicating, hyper‑gossiping primates. We leak information. So what one person knows, many people soon know. That gives you a simple intuition: the more brains, the more ideas; the more ideas, the better the technology; the better the technology, the more people the world can support. Kremer turned that into a macro‑history story: in a Malthusian agricultural world, higher productivity lets you farm smaller plots while maintaining the same standard of living. That supports a larger population. A larger population means more brains. More brains means more ideas. More ideas push technology up further. In equations, you write something like:

- The proportional rate of growth of the stock of ideas h is proportional to population P:h = \alpha P^{\sigma}

where \alpha captures how effectively brains produce ideas, and \sigma captures the “stepping on toes” effect: duplication of effort, multiple research teams reinventing the same wheel, and so on. If \sigma = 1, two heads are exactly twice as good as one. With \sigma = 0.5, two heads are only \sqrt{2} times as good—still better, but with diminishing returns to sheer headcount.

You then plug this innovation process into a very stripped‑down Malthusian model:

- Population growth n rises with living standards relative to subsistence. Write income per capita as y and the subsistence level as y_{sub}. Then you can posit something liken = \beta \frac{y - y_{sub}}{y_{sub}} + \varepsilon

where \beta is the sensitivity of population growth to higher living standards, and \varepsilon is a random “shock” (plagues, wars, good harvests, and so on).

- Productivity growth g is helped by faster idea growth h but hurt by faster population growth n, because more people pressing on a finite resource base (land, mines, forests) drive diminishing returns. A simple specification is:g = h - \frac{1}{\gamma} n

where \gamma measures how important ideas are relative to natural resources in generating prosperity.

For the purposes of this lecture, we fix some plausible‑looking parameter values:

- \sigma = 0.5: two heads are better than one, but not twice as good.

- \beta = 0.005: doubling income per capita above subsistence raises population growth by roughly 0.5 percentage points per year.

- \gamma = 2: ideas are twice as salient as resource abundance.

- Subsistence income is about 1,100 (in 2025 dollars), because that’s roughly where pre‑modern population growth appears to have been near zero.

This is an unabashedly “fundamentalist” Malthusian model: as long as you stay in the pre‑modern income range—say, not more than five times subsistence—it assumes that higher income translates mostly into more babies rather than permanently richer individuals. It deliberately omits the modern demographic transition, when rising incomes and female education drive fertility down.

The key question we push on is: with all this in place, if you calibrate \alpha and \sigma so that the model roughly matches the pre‑200 CE history of technology, population, and income, what kind of world do you get by 1775 and 2025? Do you automatically obtain something like the Industrial Revolution and the subsequent modern growth explosion? Or do you get something much more boring? To answer that, we wrap the framework in a Python simulation that marches human history forward from 3000 BCE to 2025 CE:

1. We take a data frame of “guesstimates” of world population, technology level, and income per capita at benchmark dates from −3000 to the present. Think of these as a long‑run skeletal reconstruction of global economic history.

2. We write a function ‎⁠simulatehistory()⁠ that:

 ▫ Takes as inputs: the historical data frame; \alpha, \sigma, \beta, \gamma; the subsistence income level; and the start and end years.

 ▫ For the initial year, reads off the historical values of population P_0, technology H_0, and income per capita y_0.

 ▫ Each year:

 ⁃ Draws a random population shock from a normal distribution (yes, I know Jan Steinsson would scold me for not putting in a fat lower tail with Black Death–style catastrophes).

 ⁃ Updates population growth n_t based on the gap between current income and subsistence, plus the shock.

 ⁃ Updates idea growth h_t = \alpha P_t^{\sigma}.

 ⁃ Updates productivity growth g_t as a function of h_t and n_t.

 ⁃ Rolls forward the levels of population, technology, and income per capita.

 ▫ Returns time series for:

 ⁃ Productivity growth, population growth, idea‑stock growth

 ⁃ Levels of output per capita, population, and ideas

 ⁃ The year index

3. We then write a second function to plot model simulations against our benchmark historical series over three windows:

 ▫ −3000 to 200

 ▫ −3000 to 1775

 ▫ 200 to 2025We take logs where appropriate, being careful not to ask Python to take logs of negative numbers or zero.

The simulation is stochastic: different random draws give you different paths of plagues and booms. But the underlying structure is fixed, and the parameters \alpha and \sigma are pinned down by matching pre‑200 CE technological progress.

This is crucial: we are not fitting the Industrial Revolution. We are asking what the model—which is tailored to match ancient and classical history—would naturally predict for the period after 1500 if nothing “big and new” happens in the underlying structural parameters.

4. What the Model Gets Right—and What It Misses

When we run this calibrated model from −3000 to 200, it does surprisingly well:

- Population growth tracks our historical guesstimates reasonably closely.

- The log of the technology index also tracks pretty well, by construction: we chose \alpha and \sigma to make it do so.

- Income per capita stays hovering not that far above subsistence, with some modest upward drift in the classical era.

We occasionally see random “Bronze Age dark ages” in the simulations—big negative population shocks that look a lot like the real‑world collapse of the Mycenaeans, Hittites, and others around 1200 BCE. In these episodes, technology continues to advance, but a big chunk of population disappears. Survivors enjoy larger farms and higher incomes, but civilization, in the sense of complex states and literacy, can still crash. That is exactly what historians have in mind when they write books like 1177 B.C.: The Year Civilization Collapsed.

So far so good. The two‑heads‑are‑better‑than‑one Malthusian model, plus random shocks, is good at generating periodic civilizational collapses and recoveries in the ancient world.

But where the model fails badly is the so‑called “Late Antiquity Pause” from about 200 to 800 CE. In Eurasian history, those six centuries are ugly: Roman collapse in the West, Romano‑Persian wars in the East, the fall of the Han, the Three Kingdoms and Northern and Southern Dynasties in China, Islamic conquests, and, later, the An Lushan Rebellion. Writing more or less disappears from Britain. China’s registered population drops by roughly a third—though it is unclear how much of that is actual mortality versus successful tax evasion by local strongmen.

Our simple simulation has no mechanism for a six‑century‑long global or pan‑Eurasian stagnation. The random‑shock structure gives you occasional deep dips, but they are short: think one bad century, not six. And because ideas keep accumulating, the model’s world by 800 CE has:

- A population of around 500 million, versus perhaps 200 million in actual history.

- A significantly higher technology level and modestly higher income per capita.

Reality has fewer people, less technology, and a deeper institutional and demographic scar.

4. Counterfactual Worlds Without the Late Antiquity Pause

Run the model forward to 1775, and the gap grows.

By 1775:

- Actual world population is ~800 million.

- The simulation’s population, without a late‑antiquity disaster baked in, wants to be on the order of 2.5 billion.

Technology is higher in the simulation, but not astronomically so, and income per capita remains Malthusian: maybe twice subsistence. This is still a poor agrarian world. Nearly everyone is still a peasant or a near‑peasant.

Push the same mechanism forward to 2025, again without any structural change in the parameters:

- The model’s population is ~5 billion.

- Actual population is ~10 billion.

- The model’s technology level is far, far below reality.

- The model’s income per capita is in the range of 1,500–3,000 dollars per year—early‑modern at best. There is no demographic transition (fertility is still high), no sustained modern lifespan, no mass secondary education, no escape from the Malthusian trap. Think steampunk: a world at roughly 1890s technology with five billion people crowding the land.

By contrast, in actual history:

- After about 1775 we see a marked acceleration in the growth rate of technology, associated first with the Industrial Revolution (textiles, steam, iron and steel, railroads) and later with a cascade of general‑purpose technologies—electricity, internal combustion, modern chemistry, telecommunications, computing.

- After about 1875 technology growth stabilizes at around 2% per year or more, and income per capita doubles every generation or so.

- Population rises from ~1.3 billion in 1875 to ~8 billion today, but—crucially—income per capita rises too. We get the demographic transition: fertility falls as incomes, education (especially for women), and child survival rise.

The calibrated two‑heads‑are‑better‑than‑one Malthusian model, fit to −3000 to 200, simply does not produce an escape. It gives you more people and more ideas, yes—but not enough additional ideas per person to outrun diminishing returns to land. In the model, technology growth in 2025 is perhaps 0.3% per year, not 2% plus. And that difference is everything. So if you want to explain the world we actually live in—rich, populous, and technologically explosive—you need to say that something big changed relative to the −3000 to 200 pattern. In fact, you need to say that several big things changed. 

You then naturally slice post-200 global economic history into four analytically distinct periods, each with its own puzzle:

1. The Late Antiquity Pause (ca. 200–800 CE).
Why did Eurasian civilization suffer such a long and deep multi‑century setback—political fragmentation, population loss, institutional breakdown—relative to the earlier trend? Previous collapses (e.g., Bronze Age) were more localized; this one hits both ends of Eurasia and the middle.

2. The “Commercial‑Imperial” Era (say 800/1500–1775).
After 800, and especially after 1500, technology seems to improve faster than the pre‑200 model would predict given the smaller‑than‑counterfactual population. Trade, urbanization, state formation, and cross‑Eurasian connections all accelerate. But the world remains broadly Malthusian: higher productivity mostly supports more people, not a huge jump in average living standards.

3. The Industrial Revolution Century (1775–1875).
Here we do get something qualitatively new. A cluster of innovations—mechanized textiles, steam power, railroads, modern iron and steel—launches sustained, systematic improvements in productivity, especially in manufacturing and transport. Technology growth now runs several times faster than in any previous era.

4. The Modern Economic Growth Era (post‑1875).
After about 1875, we see repeated general‑purpose technologies (GPTs) every decade or two: electricity, chemicals, internal combustion, telecom, electronics, digital computing, and more. Each GPT takes decades to diffuse and reorganize the economy around it. Technological progress becomes not a series of one‑off inventions but a continuous, institutionalized process. Productivity doubles every 30–40 years.

Any serious theory of long‑run growth has to tackle all four: the post-200 Dark Age, the acceleration in the early modern era, the Industrial Revolution cluster, and the post‑1875 explosion.

Historians and economists have proposed a long and overlapping list of mechanisms. None of them is mutually exclusive, and all of them probably mattered. The real question is: which were necessary conditions, and which merely helped at the margin? Here is a stylized shortlist, with some deepening:

(a) **Geography and Resources.** Coal close to the surface in Britain, Belgium, and parts of Germany; deep, heavy, parasite‑poor soils suited to horse and oxen plow agriculture; temperate climates with relatively low disease burdens. The ecological argument is that if Song‑dynasty China had had British‑style coal deposits and northern Europe’s soils in the Yellow River basin, the industrial breakthrough might have come there centuries earlier. See, for example, Kenneth Pomeranz, The Great Divergence (Princeton University Press (https://press.princeton.edu/books/paperback/9780691090108/the-great-divergence)) and Robert Allen, The British Industrial Revolution in Global Perspective (Oxford University Press (https://global.oup.com/academic/product/the-british-industrial-revolution-in-global-perspective-9780521687850)).

(b) **Institutions: Property Rights and Government.** States that are strong enough to suppress roving bandits but constrained enough not to become kleptocratic stationary bandits. Governments that, at least some of the time, decide that merchants and manufacturers are assets to be protected rather than sheep to be sheared. The challenge is not just to define property rights but to enforce them against both local notables and the state’s own agents. This is the world of “inclusive institutions” versus “extractive institutions.” Representative government is one possible solution, but, as the history of classical Athens shows, representative polities can also vote for ruinous wars.

(c) **Political Fragmentation and Competition.** Western Eurasia never sustains a universal empire for long. Charlemagne, Charles V, Louis XIV, Napoleon: all attempt, none succeed. Instead we get a persistent mosaic of rival states. That has at least two effects:

- It forces rulers to care about productive capacity, because they need taxable revenue to fund armies.

- It creates a competitive marketplace for ideas: if one ruler suppresses a useful idea (religious, scientific, or technological), others can adopt it. Galileo may be silenced in one jurisdiction; Copernican ideas can still flourish in another.

By contrast, in parts of eastern Eurasia, large centralized empires may be better at internal peace but also better at shutting down “dangerous” ideas. (This is a crude generalization but captures one important line of argument.)

(d) **Culture and Religion.** Max Weber’s “Protestant ethic” argument is that certain religious cultures encouraged sober work, thrift, and reinvestment, even at the cost of joyless, anxious lives. God, on this view, pre‑selects a small fraction of humanity for salvation, and worldly success is at best a sign—not a guarantee—of grace. Partying and visible luxury are bad signals. The evidence for this as a primary causal mechanism is mixed, but it probably interacted with institutions and geography. Cultural attitudes toward time, work, and risk matter.

(e) **High‑Wage Economy and Inducement Mechanisms.** After the Black Death, Western Europe has relatively high wages (lots of land per worker) and, thanks to glaciation and geology, cheap coal. That makes it profitable to substitute capital and energy for labor even with very primitive steam engines. If wages had been low and energy dear, early steam engines would never have been adopted. This is central to Allen’s story.

(f) **Colonialism, Slavery, and Core–Periphery Dynamics.** Conquest of the Americas, plantation slavery, and later the colonization or semi‑colonization of India and China transfer resources and rents to the European core, while limiting industrial development in the periphery. The core grabs high‑productivity manufacturing; the periphery is pushed toward primary products. That helps explain why the modern world is so astonishingly unequal as well as rich. It is less obvious that it explains the initial takeoff, especially since the countries most deeply embedded in colonial extraction or plantation slavery (Spain, the American South) lag in industrialization.

(g) **Human Capital and Knowledge Transmission.** The spread of literacy, numeracy, and scientific habits of mind; the institutionalization of experimental science; tinkering cultures of artisans and engineers; patent systems and scientific societies. These are all about thickening the channels through which “two heads” can actually generate and test new ideas, rather than just gossip about the old ones.

(h) **Psychological and Social Institutions.** The design of schools, firms, guilds, and professions that channel individual ambition into positive‑sum innovation rather than negative‑sum rent‑seeking. Here the “psychology” is not individual, but institutional: how are status and rewards allocated? To priests and warriors who can monopolize violence and ideology? Or to merchants, engineers, and scientists who figure out how to make the pie bigger?

(i) **Post‑Takeoff Core–Periphery Trade Geometry.** Once you have an industrial core, the geometry of trade and demand elasticity locks in inequality. Demand for manufactured goods is relatively elastic: when manufactured goods get cheaper, people buy many more of them, so the benefits of productivity growth accrue largely to producers. Demand for food and raw materials is relatively inelastic in the medium run: when agricultural productivity increases, prices fall, so gains are passed on to consumers, not producers. Thus, once industrialization starts somewhere, that “somewhere” becomes the core that reaps disproportionate gains from worldwide technological progress.

This last story does not itself explain why industrialization began in one region rather than another, but it does explain why, once begun, it created a sharp and persistent core–periphery structure.

Given all this, where do we end up?

We cannot rerun history with different parameter settings, removing one causal mechanism at a time to see what happens. We do not know which single element, if any, was truly indispensable. What we can say is that the bare‑bones two‑heads‑are‑better‑than‑one Malthusian model, calibrated to −3000 to 200, does not remotely generate the world of 2025. It predicts a crowded, poor, early‑modern planet, not one in which average income is many multiples of subsistence and technology doubles every generation.

To get from that simulated world to ours, we seem to need a stack of extra gear: coal and geography, yes, but also particular institutions; political fragmentation and competition; scientific cultures; new forms of human capital; and an especially favorable configuration of global trade and empire. Coal with no steam engine is useless. A steam engine with no tinkering culture and no natural science goes nowhere. A scientific culture without some degree of institutional protection and freedom to experiment struggles to survive. A high‑wage economy without cheap energy might induce different technologies or none at all.

The best we can say at this point is that most of these pieces, and perhaps all of them, were jointly necessary. The Industrial Revolution and the post‑1875 growth explosion were the outcome of a historically contingent constellation of geography, institutions, culture, and chance. That is why, to understand why we are so rich, so many, and so unequal—and why this outcome was anything but guaranteed—we have to start from very simple models, see where they fail, and then patiently layer on the complications that history insists on.
