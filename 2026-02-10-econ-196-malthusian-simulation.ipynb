## Modeling & Simulating a Malthusian Economy:

Write down a very simple economic model for our pre-modern Malthusian economy. It will have four equations.

The first and second equations are simply definitions:

> $ \frac{1}{y} \frac{dy}{dt}=g $
 
> $ \frac{1}{L} \frac{dL}{dt} = n $
 
 
 
 
 
 
The first says that the proportional growth rate of living standards and productivity levels—the proportional growth rate of the output per worker y variable on the left-hand side—is equal to g, g for growth. The second says that the proportional growth rate of population and the labor force variable L, L for labor, on the right-hand side is equal to n, n for numbers.

---
---

The third and fourth equations are behavioral relationships:

> $ g = h - n/\gamma $

> $ n = \beta \left( \frac{y}{y^{sub}/phi}  - 1 \right) $
 
The third says that g—the proportional growth rate of the efficiency of labor, of living standards and productivity levels—on the left-hand side, is equal to:

* the proportional rate of growth h of human ideas about technology,
* minus the proportional rate of growth of population and the labor force n
* divided by a parameter ɣ that tells us how salient ideas about technology are in generating productivity vis-à-vis resource scarcity.

It is human ingenuity versus resource scarcity. And resource scarcity is made more dire by population increases.

If we were focusing on math, we would do calculus things and see:

> $ \frac{d}{dt} \left(\frac{Y}{L}\right)^* = 0 $ whenever $ h - \frac{n}{\gamma} = 0 $

> $ n^{*mal} = \gamma h $ 

is the population growth rate at which the efficiency of labor and living standards are constant: $ \frac{d}{dt} \left(\frac{Y}{L}\right)^* = 0 $

When population is growing at the rate $ n^{*mal} $, the efficiency of labor—and thus the steady-state growth-path level of production per worker $ Y/L $—is constant. This captures the idea that even though human technology was advancing over the more than five millennia before the Industrial Revolution, living standards were not because the potential benefits from technology and organization for productivity were offset by the productivity-diminishing effects of smaller farm sizes and more costly other natural resources to feed and provide for the growing population. 

In any previous decade, this would have been an essential step. But it is not essential for us here and now.

---
---

The fourth equation says that the population growth rate variable n on the left-hand side will be such that population will grow if living standards y are above, and shrink if living standards are below, some “subsistence” level y^{sub}, and also on the fraction $ 1/\phi $ of production that is devoted to necessities, not conveniences and luxuries, and thus enters into reproductive and survival fitness.

How much? That depends on a parameter β that tells us how responsive fertility and mortality are to want and deprivation. Twice as big a gap between living standards and subsistence will produce twice as fast a population response. As people get poorer, fertility drops: women become sufficiently skinny that ovulation becomes hit-or-miss. And as people get poorer, mortality rises: it is not just that some people starve to death, it is that the malnourished have compromised immune systems, and malnourished children, especially, are easily carried off by the common cold:

> $ \frac{dL/dt}{L} = \frac{d\ln(L)}{dt} = n = \beta  \left( \frac{y}{\phi y^{sub}}-1 \right) $

Then for population to be growing at its Malthusian rate:

> $ \gamma h = \beta \left(\frac{1}{\phi}\right) \left( \frac{y}{y^{sub}}- \phi \right)  $

> $ y^{*mal} = \phi y^{sub} \left( 1 + \frac{ n^{*mal}}{\beta}\right) = \phi y^{sub} \left( 1 + \frac{ \gamma h}{\beta}\right) $

Note that this fertility behavior only holds for poor populations. When populations grow rich and literate enough—and when women acquire enough social power—human societies undergo the demographic transition: women limit their pregnancies to the number of children they desire, confident that they will pretty much all survive to outlive them. Beyond a certain income level, this no longer holds. But it did hold up until well after the start of the Industrial Revolution.

```python
import numpy as np               # math package
import matplotlib.pyplot as plt  # plotting package
import random as rnd             # random number package
```

```python
g = []                           # time-series list for productivity growth
n = []                           # time-series list for population growth
epsilon = []                     # uncorrelated random shocks to population
L = [1]                          # time-series list for population & labor force level
y = [1.25]                       # time-series list for productivity & living standard level

h = 0.0005                       # annual rate of technological progress
gamma = 2                        # salience of ideas vis-à-vis resources in determining productivity
beta = 0.025                     # annual response of population growth to relative living standards

for t in range(999):                                        # loop through time for 1000 years
    epsilon = epsilon + [rnd.normalvariate(0, 0.02)]        # calculate this year's population shock
    n = n + [beta * (y[t] - 1) + epsilon[t]]                # calculate this year's population growth rate
    g = g + [h - n[t]/gamma + rnd.normalvariate(0, 0.0005)] # calculate this year's productivity change
    y = y + [y[t]*np.exp(g[t])]                             # calculate this year's productivity level
    L = L + [L[t]*np.exp(n[t])]                             # calculate this year's population level
```

```python
tseries = range(1000)      # set up a time-series list of years for the horizontal axis

fig, ax = plt.subplots()
ax.set_xlabel("Population (L)")
ax.set_ylabel("Living standards (y)")
ax.plot(L, y)

plt.show()
```

### The Class Thinks:

**1\. Land, fixed factors, and the wage–population loop:** The story told is the strong Malthusian one. Land is treated as the central fixed factor in an agrarian economy: as population rises on a fixed land base, marginal product of labor falls, so output per worker and thus real wages decline. When wages are temporarily pushed up—by technology, luck, institutions, or good harvests—better nutrition, health, and earlier marriage raise population growth; more people then dilute the land–labor ratio and push wages back down toward subsistence.

Also strongly noted:

*   Technological or institutional improvements mostly show up as **higher population, not permanently higher wages**.
    
*   The Clark–DeLong setup is understood as a feedback loop: income → demographic response → land constraint → income again.
    

\======

**2\. Fertility vs. mortality as adjustment margins:** It is important to distinguish:

*   **Mortality:** Higher incomes improve nutrition and disease resistance, sharply lowering death rates—especially infant and child mortality. Low incomes increase vulnerability to famine, disease, and epidemics. Plagues and famines are fast, dramatic “positive checks”.
    
*   **Fertility:** Higher incomes allow earlier and more universal marriage, more births within marriage, and better maternal health; low incomes delay or prevent marriage and sometimes suppress ovulation through malnutrition. Preventive checks (late marriage, lower marital fertility) operate more slowly, often through norms.
    

On which margin “matters more,” there is a split:

*   A large number say **mortality dominates in the short run**—crises quickly reset populations.
    
*   Several (especially those thinking with Clark) stress **fertility and marriage norms as key to long‑run equilibrium living standards**: societies with strong preventive checks (late marriage) can sustain higher incomes at the Malthusian steady state.
    
*   Most end up with the nuanced view: both margins matter, but mortality drives volatility, fertility shapes the long‑run level.
    

\======

**3\. Luxuries, norms, and higher living standards without escape:** There is broad convergence to Malthus’s position: on the idea that you can raise equilibrium living standards _within_ a Malthusian world by weakening the translation of higher income into more surviving children:

*   If households devote more income to **luxury or status consumption** (better housing, clothing, food variety, comfort, leisure), less of the surplus goes into child quantity.
    
*   If norms push **later marriage, smaller families, or higher “required” standard before forming a household**, fertility at any given wage is lower.
    

This combination lets wages settle at a **higher Malthusian equilibrium**: fewer people share the same land, so those who exist consume more. But everyone grasps that this does not break the regime: the core logic—income affects demographic behavior, which interacts with a fixed factor to limit per‑capita income—remains intact.

But it is a higher plateau, not an escape ramp.

\======

**4\. Inequality: elites vs. masses and evidence problems:** On inequality:

*   Elites control land, taxation, rents, and coercive power, so they can appropriate a rising share of surplus as population grows. Mass wages are pinned near subsistence, while elite consumption (palaces, luxury goods, art, education, stature/height) can rise.
    
*   Some note that even elites face constraints (urban disease, violence), but still live vastly better than peasants.
    
*   Many explicitly warn that **archaeological and literary evidence is strongly elite‑biased**: big houses, rich burials, monumental architecture, and literary celebrations of opulence say much about the extraction machine and very little about median living standards.
    
*   The common takeaway: impressive cultural or material remains often signal **high inequality**, not broad prosperity.
    

\======

**5\. DeLong vs. Guthmann: is “Malthusian” doing too much work?:** The vibe is that I have overstated things::

*   Malthusian dynamics are powerful and pervasive in -3000 to 1500.
    
*   But I am too **totalizing**. Perhaps I treat “Malthusian” as a label that can rationalize almost any pattern—long upswings become “waves in a trap” rather than qualitatively distinct growth episodes.
    
*   Guthmann emphasizes **heterogeneity, structural change, and long efflorescences**: trade, institutions, urbanization, and state capacity that seem to raise per‑capita incomes for centuries, even if not indefinitely compounding.
    
*   The implicit advice: keep Malthusian logic as a **baseline organizing framework**, but grant Guthmann more space—acknowledging genuine, region‑specific, and temporally extended improvements in living standards, “efflorescences”—not treating all of them as mere random noise around a flat trend.
    

\======

**6\. Chatterjee–Vogl and Malthusian relevance in the next 25 years:** Many developing countries have “escaped” classic Malthusian dynamics because rising income is now associated with **lower fertility**, not higher. But perhaps there will be regressions: Possibilities to flag:

*   **Climate change and environmental degradation** reducing effective land productivity (crop failures, water stress, ecological damage), turning land and other natural resources back into binding constraints.
    
*   **Pandemics, wars, and state failure** undermining health systems, disrupting trade, and raising mortality, especially in vulnerable regions.
    
*   **Stalling of the demographic transition**: if income growth slows, education and human‑capital improvements stall, or norms shift such that fertility stops falling, then population pressure could again erode per‑capita gains.
    

There will not be a global return to a full Malthusian trap. Instead, they see Malthusian logic potentially re‑emerging as **locally powerful** in fragile states and regions: growth showing up as population pressure and resource stress rather than as durable improvements in average living standards.

\=====

The canonical Malthusian model is powereful. But be appropriately cautious about elite‑skewed evidence. And there is definitely space for a richer account focused in “efflorescences”, in which the Malthusian model is the backbone but not the entire skeleton of pre‑modern economic history.
